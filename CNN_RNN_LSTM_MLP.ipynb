```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, Flatten, Dense
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt

# Assuming your dataframe is named ds
# Features (X): date, onpromotion, oil_prices
X = ds[['date', 'onpromotion', 'oil_prices']].values
# Target variable (y): sales
y = ds['sales'].values

# Convert the date to numerical representation using label encoding
X[:, 0] = pd.to_datetime(X[:, 0]).astype(np.int64) // 10**9

# Normalize the data
scaler_X = MinMaxScaler()
scaler_y = MinMaxScaler()
X_scaled = scaler_X.fit_transform(X)
y_scaled = scaler_y.fit_transform(y.reshape(-1, 1)).reshape(-1)

# Reshape the data for CNN (assuming you want to use a window of size 3)
window_size = 3
X_cnn = np.array([X_scaled[i:i+window_size, :] for i in range(len(X_scaled) - window_size)])
y_cnn = y_scaled[window_size:]

# Create CNN model
model = Sequential()
model.add(Conv1D(filters=64, kernel_size=2, activation='relu', input_shape=(window_size, X.shape[1])))
model.add(Flatten())
model.add(Dense(1, activation='linear'))

# Compile the model
model.compile(optimizer='adam', loss='mean_squared_error')

# Train the model and store training history
history = model.fit(X_cnn, y_cnn, epochs=80, batch_size=32, validation_split=0.1, verbose=1)

# Make predictions on the training set for evaluation
y_pred_cnn_train = model.predict(X_cnn)

# Inverse transform the predictions to the original scale
y_pred_cnn_train_original = scaler_y.inverse_transform(y_pred_cnn_train).reshape(-1)

# Inverse transform the actual values to the original scale
y_train_original = ds['sales'].values[window_size:]

# Calculate and print the Mean Squared Error (MSE) on the training set
mse_cnn_train = mean_squared_error(y_train_original, y_pred_cnn_train_original)
print(f'Mean Squared Error (CNN) on Training Set: {mse_cnn_train}')

# Visualize the actual vs. predicted sales on the training set
plt.plot(y_train_original, label='Actual')
plt.plot(y_pred_cnn_train_original, label='Predicted')
plt.title('Training Set: Actual vs. Predicted Sales (CNN)')
plt.xlabel('Time')
plt.ylabel('Sales')
plt.legend()
plt.show()

# Plot loss vs. epoch
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Loss vs. Epoch (CNN)')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()
```

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt

# Assuming your dataframe is named ds
# Features (X): date, onpromotion, oil_prices
X = ds[['date', 'onpromotion', 'oil_prices']].values
# Target variable (y): sales
y = ds['sales'].values

# Convert the date to numerical representation using label encoding
X[:, 0] = pd.to_datetime(X[:, 0]).astype(np.int64) // 10**9

# Normalize the data
scaler_X = MinMaxScaler()
scaler_y = MinMaxScaler()
X_scaled = scaler_X.fit_transform(X)
y_scaled = scaler_y.fit_transform(y.reshape(-1, 1)).reshape(-1)

# Create MLP model
model_mlp = Sequential()
model_mlp.add(Dense(64, activation='relu', input_dim=X.shape[1]))
model_mlp.add(Dense(32, activation='relu'))
model_mlp.add(Dense(1, activation='linear'))

# Compile the model
model_mlp.compile(optimizer='adam', loss='mean_squared_error')

# Train the model and store training history
history = model_mlp.fit(X_scaled, y_scaled, epochs=80, batch_size=32, validation_split=0.1, verbose=1)

# Make predictions on the training set for evaluation
y_pred_mlp_train = model_mlp.predict(X_scaled)

# Inverse transform the predictions to the original scale
y_pred_mlp_train_original = scaler_y.inverse_transform(y_pred_mlp_train).reshape(-1)

# Inverse transform the actual values to the original scale
y_train_original = ds['sales'].values

# Calculate and print the Mean Squared Error (MSE) on the training set
mse_mlp_train = mean_squared_error(y_train_original, y_pred_mlp_train_original)
print(f'Mean Squared Error (MLP) on Training Set: {mse_mlp_train}')

# Visualize the actual vs. predicted sales on the training set
plt.plot(y_train_original, label='Actual')
plt.plot(y_pred_mlp_train_original, label='Predicted')
plt.title('Training Set: Actual vs. Predicted Sales (MLP)')
plt.xlabel('Time')
plt.ylabel('Sales')
plt.legend()
plt.show()

# Plot loss vs. epoch
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Loss vs. Epoch (MLP)')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()
```
